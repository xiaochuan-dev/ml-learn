```python
# 全连接层计算逻辑

import torch
import torch.nn.functional as F

# 下面是一个 输入特征为1，输出特征为4的线性层
# F.linear 会自动转置weight

# weight (4,1) 的tensor
weight = torch.tensor([[1], [2], [4], [5]], dtype=torch.int32)

# bias (4,)的tensor, 即一个4个元素的向量
bias = torch.tensor([1, 3, 4, 5], dtype=torch.int32)


x1 = torch.tensor([[2], [4]], dtype=torch.int32)
out1 = F.linear(x1, weight=weight, bias=bias)
print('x1 out', out1)
print('x1 shape', x1.shape, ' to out1 shape', out1.shape)


x2 = torch.tensor([3], dtype=torch.int32)
out2 = F.linear(x2, weight=weight, bias=bias)
print('x2 out', out2)
print('x2 shape', x2.shape, ' to out2 shape', out2.shape)

x3 = torch.tensor([[[1], [2], [2], [3]], [[3], [1], [4], [3]]], dtype=torch.int32)
out3 = F.linear(x3, weight=weight, bias=bias)
print('x3 out', out3)
print('x3 shape', x3.shape, ' to out3 shape', out3.shape)
```

```python
# 张量乘法 就是多维度的矩阵乘法，只看最后两个维度，前面的维度是最终形状维度，和后面矩阵乘法得到的维度构成最终维度


x6 = torch.empty((32, 5, 6, 7), dtype=torch.int32)
x7 = torch.empty((32, 5, 7, 8), dtype=torch.int32)

x8 = x6.matmul(x7)
# 或者
x9 = x6 @ x7

print(x8.shape)
```
