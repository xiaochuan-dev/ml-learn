```python
# 全连接层计算逻辑

import torch
import torch.nn.functional as F

x = torch.tensor([[2], [4]], dtype=torch.int32)
weight = torch.tensor([[1], [2], [4], [5]], dtype=torch.int32)
bias = torch.tensor([[1, 3, 4, 5]], dtype=torch.int32)


# F.linear 会自动转置weight
out = F.linear(x, weight=weight, bias=bias)

print(out)
```
