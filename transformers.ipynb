```python
import torch, torch.nn as nn
m = nn.Transformer(d_model=8, nhead=2, num_encoder_layers=2, num_decoder_layers=2, batch_first=True)
src = torch.randint(0, 10, (2, 5, 8)).float()
tgt = torch.randint(0, 10, (2, 3, 8)).float()
out = m(src, tgt)
print(out.shape)
```

$$
Y = Attention(Q,K,V) = softmax\left[\frac{QK^{\mathrm{T}}}{\sqrt{D_k}}\right]V
$$

$QK^{\mathrm{T}}$是相似度矩阵

每个词用自己的 Query 去和所有词的 Key 算点积（相似度）→ 缩放 → softmax → 用得到的权重去加权所有词的 Value。 这就是注意力的向量版本。

```python
import torch
from torch.nn.functional import scaled_dot_product_attention

batch_size, seq_len, d_k = 1, 2, 3

Q = torch.tensor([[[1, 2, 3], [4, 5, 6]]]).float()
K = torch.tensor([[[2, 3, 4], [5, 6, 7]]]).float()
V = torch.tensor([[[3, 4, 5], [6, 7, 8]]]).float()

output = scaled_dot_product_attention(Q, K, V)

print(output)
```
